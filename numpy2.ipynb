```python
import numpy as np
```

```python
a = np.array([1,2,3])

b = np.array([4,5,6])

print(a+b)
```

```python
print(a-b)
```

```python
print(a*b)
```

```python
print(a/b)
```

```python
print(a**3) # square
```

```python
print(np.sqrt(a)) # to find the sq root
```

```python
print(np.power(a,2)) # for n exponent
```

```python
arr = np.array([10,20,30,40,50]) # defining 1d array
```

```python
print(np.mean(arr)) # to find the average
```

```python
print(np.median(arr)) # to find median value
```

```python
print(np.max(arr)) # highest value
```

```python
print(np.min(arr)) # lowest value
```

```python
print(np.sum(arr)) # total value
```

```python
print(np.std(arr)) # standard deviation
```

```python
print(np.var(arr)) # variance
```

```python
arr = np.array([1,2,3,4,5,6])

print(arr.reshape(2,3))
```

```python
arr = np.array([[1,2],[3,4]])

print(arr.flatten())
```

```python
arr = np.array([[1,2,3],
                [4,5,6]])

print(arr.T)
```

```python
np.zeros((3,3))
```

```python
np.ones((2,4))
```

```python
np.eye(4)
```

```python
np.arange(1,11) # array in particular range
```

```python
np.random.rand(5) # give random numbers below 1
```

```python
np.random.randint(1,100,10) # give random integers
```

```python
arr = np.array([5,2,9,1])

print(np.sort(arr))
```

```python
arr = np.array([1,2,2,3,3,4])

print(np.unique(arr))
```

```python
a = np.array([1,2,3])

b = np.array([4,5,6])

print(np.concatenate((a,b)))
```

```python
arr = np.array([1,2,3,4,5,6])

print(np.split(arr,3))
```
