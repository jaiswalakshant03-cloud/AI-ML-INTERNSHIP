```python
num1=int(input("enter number: "))
num2=int(input("enter number: "))
choice=input("enter operation(+,-,*,/):")
if (choice =="+"):
  print("result=",num1+num2)
elif(choice=="-"):
  print("result=",num1-num2)
elif(choice=="*"):
  print("result=",num1*num2)
elif(choice=="/"):
  print("result=",num1/num2)
else:
  print("wrong choice")
```
